#OPERATORS


1.Arthimatice Operator

```python
#Arthimatice operators
print(5+3)
print(5-9)
print(4*3)
print(20/4)
print(40/6)
print(40//6)  # gives floor value
print(10%6)   # modulus
print(2**3)   # power

```

2.Assignment Operator

```python
x = 10
print(x)
x +=2
print(x)
x -=4
print(x)
```

```python
x = 10
x *=2
print(x)
x /=5
print(x)
```

```python
x = 20
x //=5
print(x)
```

3.Comparison Operator

```python
print(10>5)
print(4<2)
print(10>=10)
print(5<=6)
print(45==45)
print(4!=4)
```

4.Logical Operator

```python
print(4>3 and 6<10)
print(54>34 and 23<10)
print(4>3 or 6<10)
print(54>34 or 23<10)
print(not(4>10))
print(not(4!=5))
```

5. Membership Operator

```python
print("a" in "akash")

print("y" not in "akash")
```
